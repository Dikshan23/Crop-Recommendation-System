```python

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from src.train import DecisionTreeCART
```

```python
df = pd.read_csv("../data/processed/processed_data.csv")

df.head()
```

```python
X = df.drop("label", axis=1)
y = df["label"]
```

```python
X_train_val, X_test, y_train_val, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
# 70% train and 10% validation
X_train, X_val, y_train, y_val = train_test_split(
    X_train_val,
    y_train_val,
    train_size=0.875,      # 70/80 = 0.875
    random_state=42,
    stratify=y_train_val
)
```

```python
print("Training samples  :", len(X_train))
print("Validation samples:", len(X_val))
print("Testing samples   :", len(X_test))
```

```python
from sklearn.tree import DecisionTreeClassifier

entropy_tree = DecisionTreeClassifier(
    criterion="entropy",
    random_state=42
)

entropy_tree.fit(X_train, y_train)
```

```python
from sklearn.metrics import accuracy_score

train_pred = entropy_tree.predict(X_train)

train_accuracy = accuracy_score(y_train, train_pred)

print(f"Training Accuracy: {train_accuracy:.2%}")
```

```python
from sklearn.metrics import accuracy_score

val_pred = entropy_tree.predict(X_val)

val_accuracy = accuracy_score(y_val, val_pred)

print(f"Validation Accuracy: {val_accuracy:.4f}")
```

```python
test_pred = entropy_tree.predict(X_test)

test_accuracy = accuracy_score(y_test, test_pred)

print(f"Test Accuracy: {test_accuracy:.4f}")
```

```python
DTmodel = DecisionTreeCART(
    max_depth=15,
    min_samples_split=10,
    min_samples_leaf=5,
    random_state=42
)

DTmodel.fit(
    X_train.values,
    y_train.values
)
```

```python
train_pred = DTmodel.predict(X_train.values)

train_accuracy = accuracy_score(
    y_train,
    train_pred
)

print(f"Training Accuracy: {train_accuracy:.2%}")
```

```python
val_pred = DTmodel.predict(X_val.values)

val_accuracy = accuracy_score(
    y_val,
    val_pred
)

print(f"Validation Accuracy: {val_accuracy:.2%}")
```

```python
test_pred = DTmodel.predict(X_test.values)

test_accuracy = accuracy_score(
    y_test,
    test_pred
)

print(f"Test Accuracy: {test_accuracy:.2%}")
```
